```python
from sklearn.datasets import load_breast_cancer
import numpy as np
```

```python
def explorer_dataset(filter=None):
    breast_cancer_dataset = load_breast_cancer()
    X, y = breast_cancer_dataset.data, breast_cancer_dataset.target
    if filter is not None:
        class_index = y == filter
        X = X[class_index]
        y = y[class_index]
    print(f"Lignes, colonnes : {X.shape}")
    classes, counts = np.unique(y, return_counts=True)
    total = len(y)
    for class_index, count in zip(classes, counts):
        percentage = count/total*100
        print(
            f"Class {class_index} "
            f"({breast_cancer_dataset.target_names[class_index]}) : {count} cases "
            f"({percentage:.2f}%)"
        )
```

```python
explorer_dataset()
explorer_dataset(0)
explorer_dataset(1)
```
